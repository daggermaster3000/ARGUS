```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Simulate data
# -----------------------------
np.random.seed(42)

n = 200

# Number of nuclei per sample
nuclei = np.random.randint(50, 500, n)

# Simulate a range of ciliation rates
# Here we assume roughly 20-100% of nuclei are ciliated
ciliation_rate = np.random.uniform(0.2, 1.0, n)

# Number of cilia
cilia = np.round(nuclei * ciliation_rate).astype(int)

# Recalculate the actual rate after rounding
ciliation_rate = cilia / nuclei


# -----------------------------
# 3D plot
# -----------------------------
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

scatter = ax.scatter(
    cilia,
    nuclei,
    ciliation_rate,
    s=40,
    alpha=0.7
)

ax.set_xlabel("Number of cilia")
ax.set_ylabel("Number of nuclei")
ax.set_zlabel("Ciliation rate (cilia / nuclei)")

ax.set_title("Cilia vs nuclei vs ciliation rate")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(
    nuclei,
    cilia,
    c=ciliation_rate,
    cmap="viridis",
    s=50,
    alpha=0.8
)

plt.colorbar(label="Ciliation rate (cilia / nuclei)")

plt.xlabel("Number of nuclei")
plt.ylabel("Number of cilia")
plt.title("Cilia count vs nuclei count")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Example datasets
nuclei = np.array([20, 40, 100, 200, 500])
cilia = np.array([10, 20, 50, 100, 250])

ratio = cilia / nuclei

fig, ax = plt.subplots(figsize=(8, 6))

# Plot absolute counts
ax.scatter(
    nuclei,
    cilia,
    s=150,
    zorder=3
)

# Connect each point to origin
for x, y, r in zip(nuclei, cilia, ratio):
    ax.plot(
        [0, x],
        [0, y],
        alpha=0.3
    )
    
    ax.text(
        x,
        y + 10,
        f"{y}/{x} = {r:.2f}",
        ha="center",
        fontsize=11
    )

# Ratio reference lines
x = np.linspace(0, 550, 100)

for r in [0.25, 0.5, 0.75, 1.0]:
    ax.plot(
        x,
        r * x,
        linestyle="--",
        alpha=0.4
    )
    ax.text(
        510,
        r * 510,
        f"ratio = {r}",
        va="center"
    )

ax.set_xlabel("Number of nuclei")
ax.set_ylabel("Number of cilia")

ax.set_title(
    "A ratio measures the relationship between two quantities"
)

ax.set_xlim(0, 550)
ax.set_ylim(0, 550)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

n = 300

# Simulated nuclei counts
nuclei = np.random.randint(20, 1000, n)

# Assume true ciliation probability = 70%
true_rate = 0.70

# Simulate cilia as a binomial sampling process
cilia = np.random.binomial(nuclei, true_rate)

# Calculate observed ratio
rate = cilia / nuclei

plt.figure(figsize=(8, 6))

plt.scatter(
    nuclei,
    rate,
    alpha=0.5
)

plt.axhline(
    true_rate,
    linestyle="--",
    label="True rate = 0.70"
)

plt.xlabel("Number of nuclei")
plt.ylabel("Ciliation rate (cilia / nuclei)")
plt.title("Ciliation rate as a function of sample size")

plt.ylim(0, 1)
plt.legend()

plt.tight_layout()
plt.show()
```
